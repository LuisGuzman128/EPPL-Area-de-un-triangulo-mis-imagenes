## Fórmula para calcular el Área
$A = \frac{base \times altura}{2}$
---
## Datos de entrada
- ¿Base ?  
- ¿Altura ?
---

```python
# calcular area y perimetro de un triangulo
# variables A, P, a, b, c
# formulas A = √[s(s-a)(s-b)(s-c)]   y   P = a + b + c
# s = (a + b + c)/2

import math

print("calculo de area y perimetro de un triangulo")
a = float(input("ingrese el valor del lado a: "))
b = float(input("ingrese el valor del lado b: "))
c = float(input("ingrese el valor del lado c: "))

s = (a + b + c) / 2
A = math.sqrt(s * (s - a) * (s - b) * (s - c))
P = a + b + c

print("el area del triangulo es: ", A)
print("el perimetro del triangulo es: ", P)
print("Axel Guzman NL = 23")
```
