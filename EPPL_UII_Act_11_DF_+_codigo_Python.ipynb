## Fórmula para calcular el Área
$A = \frac{base \times altura}{2}$
---
## Datos de entrada
- ¿Base ?  
- ¿Altura ?
---
## Diagrama de flujo en LucidChart / tldraw
![Diagrama de flujo Área del Triángulo](https://raw.githubusercontent.com/LuisGuzman128/EPPL-Area-de-un-triangulo-mis-imagenes/refs/heads/main/diagrama%20de%20flujo.PNG)




```python
# calcular area y perimetro de un triangulo
# variables A, P, base, altura, l1, l2, l3
# formulas A = (base * altura) / 2    y    P = A + B + C

print("calculo de area y perimetro de un triangulo")

base = float(input("ingrese el valor de la base: "))
altura = float(input("ingrese el valor de la altura: "))
A = float(input("ingrese el valor de A: "))
B = float(input("ingrese el valor de B: "))
C = float(input("ingrese el valor de C: "))

A = (base * altura) / 2
P = A + B + C

print("el area del triangulo es: ", A)
print("el perimetro del triangulo es: ", P)
print("Axel Guzman NL = 23")
```
